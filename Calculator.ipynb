```python
def calculate_expression(expr):
    tokens = expr.split()

    # نحول الأرقام لـ float
    tokens = [float(t) if t not in "+-*/" else t for t in tokens]

    # أول حاجة: الضرب والقسمة
    i = 0
    while i < len(tokens):
        if tokens[i] == '*':
            result = tokens[i-1] * tokens[i+1]
            tokens[i-1:i+2] = [result]
            i = 0
        elif tokens[i] == '/':
            if tokens[i+1] == 0:
                return "Error: Division by zero"
            result = tokens[i-1] / tokens[i+1]
            tokens[i-1:i+2] = [result]
            i = 0
        else:
            i += 1

    # بعد كده: الجمع والطرح
    result = tokens[0]
    i = 1
    while i < len(tokens):
        if tokens[i] == '+':
            result += tokens[i+1]
        elif tokens[i] == '-':
            result -= tokens[i+1]
        i += 2

    return result


# البرنامج الرئيسي
while True:
    expr = input("Enter expression (or q to quit): ")

    if expr.lower() == 'q':
        print("Mission has Ended :) ")
        break

    try:
        print("Result =", calculate_expression(expr))
    except:
        print("Invalid input! Use spaces like: 2 + 3 * 4")
```
